```python
# 19. callable() and __call__()
# Assignment: Create a class Multiplier with an __init__() to set a factor. Define a __call__() method that multiplies an input by the factor. Test it with callable() and by calling the object like a function.

class Multiplier:
  def __init__(self, factor):
    self.factor = factor

  def __call__(self, x):
    return x * self.factor

# Create a Multiplier object with a factor of 5
multiplier = Multiplier(5)
# Using callable() to check if the object is callable
print(callable(multiplier))

result = multiplier(10) # # Equivalent to calling multiplier.__call__(10)
print(result)
```
