```python
# 2. Using cls
# Assignment:
# Create a class Counter that keeps track of how many objects have been created. Use a class variable and a class method with cls to manage and display the count.

class Counter:
  # Class variable: sab objects ke liye common hai
  object_count = 0

  def __init__(self):
    # Jab bhi new object banta hai, count ko 1 barhaya jata hai
    Counter.object_count += 1

  @classmethod
  def display_count(cls):
    print(f'Total object created: {cls.object_count}')

# 4 new objects bana rahe hain
object1 = Counter()
object2 = Counter()
object3 = Counter()
object4 = Counter()
# Class method se object count display karwa rahe hain
Counter.display_count()


# Output:
#  Total object created: 4
```
