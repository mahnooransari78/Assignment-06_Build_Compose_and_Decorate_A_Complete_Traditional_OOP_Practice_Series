```python
# 21. Make a Custom Class Iterable
# Assignment: Create a class Countdown that takes a start number. Implement __iter__() and __next__() to make the object iterable in a for-loop, counting down to 0.

class countdown:
  def __init__(self, start):
    self.start = start  # Initialize the start number
    self.current = start  # Set the current value to start


  def __iter__(self):
    return self   # The iterator object is the same as the iterable object

  def __next__(self):
    if self.current <= 0:
      raise StopIteration  # # Stop iteration when the countdown reaches 0
    else:
      self.current -= 1    # Decrement the current value by 1
      return self.current  # Return the next value in the countdown

countdown = countdown(5)
# Iterate through the countdown using a for-loop
for number in countdown:
  print(number)
```
