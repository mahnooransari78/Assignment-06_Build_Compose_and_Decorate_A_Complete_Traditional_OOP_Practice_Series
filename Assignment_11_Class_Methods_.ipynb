```python
# 11. Class Methods
# Assignment: Create a class Book with a class variable total_books. Add a class method increment_book_count() to increase the count when a new book is added.

class Book:
  # Class variable — sab objects ke liye common
  total_books = 0

  def __init__(self, title):
    self.title = title
    # Jab bhi naya book object banayen, count barhao
    Book.increment_book_count()

  @classmethod
  def increment_book_count(cls):
      cls.total_books += 1

  @classmethod
  def display_total_books(cls):
      print(f'Total books: {cls.total_books}')


book1 = Book("Python Basics")
book2 = Book("Python Advance")
book3 = Book("OOP in Python")
book4 = Book("Data Science 101")

# Show total count
Book.display_total_books()


```
