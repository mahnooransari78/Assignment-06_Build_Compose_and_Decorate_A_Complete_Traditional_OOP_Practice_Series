```python
# 13. Composition
# Assignment: Create a class Engine and a class Car. Use composition by passing an Engine object to the Car class during initialization. Access a method of the Engine class via the Car class.

# Composition ka matlab hota hai:
# Ek class ke andar doosri class ka object use karna.

# Engine class
class Engine:
  def start_engine(self):
    print("Engine started...")
# Car class using composition
class Car:
  def __init__(self, engine):
    self.engine = engine  # Composition: Car HAS an Engine

  def start(self):
    print("Car is starting...")
    self.engine.start_engine() # Engine's method used here
# Engine object
my_engine = Engine()
# Car object, engine injected via constructor
my_car = Car(my_engine)
# Car starts, which uses Engine internally
my_car.start()
```
