```python
# 17. Class Decorators
# Assignment: Create a class decorator add_greeting that modifies a class to add a greet() method returning "Hello from Decorator!". Apply it to a class Person.


def add_greeting(cls):
  def greet(self):
    return "Hello from Decorator!"

  setattr(cls, 'greet', greet)  #setattr(cls, 'greet', greet) adds a method greet() to the class at runtime.

  return cls

@add_greeting
class Person:
  def __init__(self, name):
    self.name = name

p = Person("Ali")
print(p.name)
print(p.greet())
```
