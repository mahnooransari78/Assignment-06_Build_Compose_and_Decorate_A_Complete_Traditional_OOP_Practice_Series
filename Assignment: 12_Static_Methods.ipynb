```python
# 12. Static Methods
# Assignment: Create a class TemperatureConverter with a static method celsius_to_fahrenheit(c) that returns the Fahrenheit value.


class TemperatureConverter:

  @staticmethod
  def celsius_to_fahernheit(c):
    return (c * 9/5) + 32

# Use static method without creating an object
temp_c = 37
temp_f = TemperatureConverter.celsius_to_fahernheit(temp_c)
print(f'{temp_c}c = {temp_f}F')
```
