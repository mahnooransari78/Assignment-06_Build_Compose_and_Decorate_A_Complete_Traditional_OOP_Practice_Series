```python

class Bank:
# Class variable — sab objects ke liye common
  bank_name = "Old Bank"

  def __init__(self, customer_name):
    self.customer_name = customer_name

  def show_details(self):
    print(f'Customer: {self.customer_name}, Bank: {Bank.bank_name}')

   # Class method to change bank name
  @classmethod
  def chang_bank_name(cls, name):
      cls.bank_name = name

# Create 2 customers (objects)
customer1 = Bank("Ali")
customer2 = Bank("sara")

# Pehle dono ka bank name same hoga
customer1.show_details() # Customer: Ali, Bank: Old Bank
customer2.show_details() # Customer: sara, Bank: Old Bank

# Ab class method se bank name change karte hain
Bank.chang_bank_name("My bank")

# Dubara show karte hain — dono ka naam update ho chuka hoga
customer1.show_details()  # Customer: Ali, Bank: My bank
customer2.show_details()  # Customer: sara, Bank: My bank
```
